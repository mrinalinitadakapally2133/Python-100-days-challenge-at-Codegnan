```python
# read list of number and if number is even add 0.5 and if number is odd add 1
# lst = [1,2,3,4,5]
# res = [2, 2.5, 4 ,4.5 ,6]
lst = list(map(int, input().split()))
for i in range(0, len(lst)):
    if lst[i] % 2;
        lst[i] = lst[i] + 1
        else;
        lst[i] += 0.5
print(lst)
```
