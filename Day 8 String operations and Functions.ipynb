### string functions
- Case conversion functions - lower(), upper(), title(), swapcase(), capitalize()


```python
s="I am a python BatcH student"
print(s.lower()) #string function, whereas len(s)) #building/pre-defined function
print(s.upper())
print(s.title())
print(s.swapcase())
print(s.capitalize())
```

```python
#sub string index finding
#find()
s="I am a python batch student"
print(s.find("o"))
```

```python
#sub string index finding
#find()
s="I am a python batch student"
print(s.find("t")) #to know about function description press Shift_Tab placing the cursor after the opening "("
```

```python
#sub string index finding
#find()
s="I am a python batch student"
print(s.find("t", 10)) #starts checking from 10th position
```

```python
#strip() - it removes the white space
s="     codegnan     "
print(s)
print(s.strip()) #it removes white space at both sides of strip
print(s.lstrip()) #it removes white space at left side of strip 
print(s.rstrip()) #it removes white space at right side of strip
```

```python
s="@@@@@cod@@eganan@@@@@"
print(s)
print(s.strip("@"))
print(s.lstrip("@")) 
print(s.rstrip("@"))
```

```python
#split(sep) - it returns the list of substring and it seperates the string with sep
s="@@@@@cod@ @egnan@@@@@"
print(s)
print(s.split()) 
print(s.split("@"))  
print(s.split("n"))
```

```python
s="I am a python batch student"
print(s.split())
```

```python
#s.find(): Returns index of first occurrence of substring. Returns -1 if not found.
#s.index(): Returns index of first occurrence of substring. Raises ValueError if not found.
s = "codegnan"
print(s.find("e"))
print(s.find("z"))  
print(s.index("e"))  
```

```python
#case checking functions
s1="CODE"
s2="gnan"
s3="123"
s4="abc123" #even if there's space,@ it gives false
s5="Codegnan"
print(s1.isupper()) #is upper
print(s2.isupper())
print(s1.islower()) #is lower
print(s2.islower()) 
print(s3.isdigit()) #is digit
print(s4.isdigit())
print(s4.isalnum()) #is all num
print(s5.isalpha()) #is alphabet
```
