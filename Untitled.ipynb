```python
# who is topper

print("Welcome to the who is topper")
n = int(input("Enter the number of student in the class:"))
names = []
marks = []
total_marks = []
percentage = []
print("Enter the all student marks")
for i in range(n):
    print(f"Enter the student {i+1}  marks")
    name = input("Enter the student name")
    print("Enter the telugu, english, hindi, maths, science, social marks, as per the order")
    telugu, english, hindi, maths, science, social = map(int, input().split()[:6])
    class_marks= telugu + english + hindi + science + social
    student_percentage = (total_marks/600) * 100
              # store all values in list
    names.append(name)
    marks.append([telugu, english, hindi, maths, science, social])
    total_marks.append(class_marks)
    percentage.append(student_percentage)
# print the topper details
print("Topper in the class is")
top_marks = max(total_marks)
toppers_pos = []

for i in range(n):
    if total_marks[i] ==top_marks:
        toppers_pos.append(i)
# print topper details
for i in range (len(toppers_pos)):
    print("Name of the student is:", name[i])
    print("Marks of the student is:",marks[i])
              
# each student marks
for i in range(n):
    print("Name of the student is:", name[i], "and marks is:",marks[i])
              
              
              
              
```
