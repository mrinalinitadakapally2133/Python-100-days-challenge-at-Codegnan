```python
#username validation
    -minimum length=8
    -atleast ione uppercase and one lowercase letter
    -special characters("_" and "@")
    -it contains digits
#password validation
```

```python
s=input() #Mrinalini@123
validate_string="qwertyuioplkjhgfdsazxcvbnmQWERTYUIOPLKJHGFDSAZXCVBNM1234567890@_"
if len(s)>=8:
    #check atleast one uppercase and one lowercase letter
        #-first character is upper or lower
    if s[0].isupper() or s[0].islower():
        for i in range(1,len(s)):
            if s[i] not in validate_string:
                print(f"Invalid username because character {s[i]}")
                break
        else:
            print("Valid string") 
        

        #-special characters("_","@")
        #-it contains digits
    else: 
        print("username not valid because first character is not upper or lower")
else:
    print("username not valid because length is lesser than 8")
    
    
```
